* **Module:** A module is a Python file containing reusable functions, classes, variables, and code.
* **Import statement:** The `import` statement is used to include and use a module in a Python program.
* **Math module:** The `math` module provides built-in mathematical functions and constants such as `sqrt()`, `factorial()`, and `pi`.


```python
import math # include maths module in your code using import statement
print(dir(math))  # list down all functions in maths module
# help(math)  # get detailed documentation
```

```python
# print square root of a number

import math

# take input from user
num=float(input("enter a number: "))

# find square root
print("square root =",math.sqrt(num))
```

```python
# calculate exponent(power) of a number

import math

# take base and power
num=float(input("enter base: "))
power=float(input("enter power: "))

# calculate power
print("answer =",math.pow(num,power))
```

```python
# square root, square, cube, ceil and floor
import math

# take input
num=float(input("enter a number: "))

# perform math operations
print("square root =",math.sqrt(num))
print("square =",math.pow(num,2))
print("cube =",math.pow(num,3))
print("ceil =",math.ceil(num))
print("floor =",math.floor(num))
```

```python
# factorial and absolute value
import math

# take input
num=int(input("enter a positive number: "))

# find factorial
print("factorial =",math.factorial(num))

NUM2 = int(input("enter a number: "))
# find absolute value
print("absolute value =",math.fabs(NUM2))
```

```python
# gcd (hcf) and lcm
import math

# take two numbers
a=int(input("enter first number: "))
b=int(input("enter second number: "))

# find gcd and lcm
print("gcd =",math.gcd(a,b))
print("lcm =",math.lcm(a,b))
```

# LOGARITHM AND TRIGONOMETRIC FUNCTIONS

```python
import math

# take input
num=float(input("enter a number: "))

# find logarithms
print("natural log =",math.log(num))
print("log base 10 =",math.log10(num))
print("log base 2 =",math.log2(num))
```

```python
import math

# take angle in degrees
angle=float(input("enter angle in degrees: "))

# convert degree to radian
radian=math.radians(angle)

# calculate trigonometric values
print("sin =",math.sin(radian))
print("cos =",math.cos(radian))
print("tan =",math.tan(radian))
```

# calulate area of a circle using math module functions

```python
import math

# take radius
r=float(input("enter radius: "))

# calculate area
area=math.pi*math.pow(r,2)
print("area =",area)
```

```python
# surface area of a cylinder

import math

# take radius and height
r=float(input("enter radius: "))
h=float(input("enter height: "))

# calculate area
area=2*math.pi*r*(r+h)

print("surface area =",area)
```

```python
# volume of sphere
import math

# take radius
r=float(input("enter radius: "))

# calculate volume
volume=(4/3)*math.pi*math.pow(r,3)

print("volume =",volume)
```

```python
# volume of cylinder
import math

# take radius and height
r=float(input("enter radius: "))
h=float(input("enter height: "))

# calculate volume
volume=math.pi*math.pow(r,2)*h

print("volume =",volume)
```

# DISTANCE BETWEEN TWO POINTS

image.png

```python
import math

# take coordinates
x1=float(input("enter x1: "))
y1=float(input("enter y1: "))
x2=float(input("enter x2: "))
y2=float(input("enter y2: "))

# calculate distance
distance=math.sqrt(math.pow(x2-x1,2)+math.pow(y2-y1,2))

print("distance =",distance)
```

```python
import math

# take input
p=float(input("enter principal: "))
r=float(input("enter rate: "))
t=float(input("enter time: "))

# calculate compound interest
amount=p*math.pow(1+r/100,t)
interest=amount-p

print("amount =",amount)
print("compound interest =",interest)
```
