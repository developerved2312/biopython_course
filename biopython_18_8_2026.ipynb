# While Loop in Python

<h2> A while loop is used when you want to repeatedly execute a block of code as long as a given condition is True.  </h2>


```python
# Check condition → Execute code → Update condition → Repeat

#SYNTAX
while condition:
    # Code to execute
    statement
```

while is the keyword used to create the loop.

condition is a Boolean expression that evaluates to True or False.

The indented statements execute repeatedly while the condition remains True.

Python uses indentation to identify the body of the loop.


# main difference

A while loop is generally useful when the number of iterations is not known beforehand.

A for loop is commonly used when iterating over a known sequence or range

```python
# Write a Python program to print numbers from 1 to 10 using a while loop.

# Initialize the counter
i = 1
# Repeat while i is less than or equal to 10
while i <= 10:
    print(i,end=",")
    # Increment the counter
    i = i + 1  # or you can write i += 1 it has a similar logic
```

```python
# Write a Python program to print numbers from 10 to 1 using a while loop.

# Initialize the counter
i = 10
# Repeat until i becomes less than 1
while i >= 1:
    print(i,end=", ")
    # Decrease the counter
    i = i - 1
```

```python
# Print even numbers from 2 to 10

# we will start from number 2
i = 2
# Continue until 10
while i <= 10:
    print(i,end=",")
    # Iincrease by 2 to obtain next even number like 2, 2+2=4, 4+2=6 and so on...
    i = i + 2
```

```python
# Multiplication table by taking input from user
n = int(input("Enter a number: "))
i = 1  # start from 1
# Print 10 multiples
while i <= 10:
    print(n * i)
    # Move to the next multiple
    i = i + 1
```

```python
# Program to repeatedly ask for a password

password = ""

while password != "python123":
    password = input("Enter password: ")

print("Login successful!")
```

```python
# Program to keep accepting numbers until the user enters 0

number = int(input("Enter a number: "))

while number != 0:
    print("You entered:", number)

    # Ask for the next number
    number = int(input("Enter a number: "))

print("Loop ended because you entered 0.")
```

# Loop control statements in Python alter the standard execution flow of for and while loops. They allow you to terminate a loop early, skip specific iterations.

image.png

```python
# BREAK: stop the loop in between

for number in range(1, 10):
    if number == 5:
        break  # Exit the loop completely
    print(number)

print("Loop finished!")

"""
The break statement terminates the loop entirely and shifts program control
to the first line of code immediately following the loop block.
When to use: Use this when you have found a specific target value or met a
critical exit condition, making further iterations unnecessary.
"""
```

```python
# continue - skip some iteration

for number in range(1, 6):
    if number == 3:
        continue  # Skip printing 3, move to 4
    print(number)
"""
continue statement skips the remaining code inside the current iteration and
forces the program to jump directly to the next cycle of the loop.
When to use: Use this to bypass certain elements
"""
```

```python
# PASS: The pass statement is a null operation; it does absolutely nothing when executed
for number in range(1, 4):
    if number == 2:
        pass  # Placeholder: do nothing for 2 right now
    print(number)

# USE CASE: WHEN YOU WANT AN EMPTY LOOP YOU WILL DEFINE LATER
```

# Python's match-case is useful when you want to compare one value against multiple possible patterns. It is similar to switch-case in languages such as C++ and Java.

image.png

```python
# Simple match-case example

day = int(input("Enter day number (1-7): "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid day number")
```

```python
# Simple calculator using match-case

operator = input("Enter operator (+, -, *, /): ")

num1 = float(input("enter first number: "))
num2 = float(input("enter second number: "))

match operator:

    case "+":
        print(num1 + num2)

    case "-":
        print(num1 - num2)

    case "*":
        print(num1 * num2)

    case "/":
        print(num1 / num2)

    case _:
        print("Invalid operator")
```

```python
# Check whether a number is positive, negative, or zero
# using match-case

number = int(input("Enter a number: "))

match number:

    # Check for zero
    case 0:
        print("The number is zero.")

    # Positive number
    case n if n > 0:
        print("The number is positive.")

    # Negative number
    case n if n < 0:
        print("The number is negative.")
```
