Write a program that accepts the three sides, a, b, and c, of a triangle, calculates its three angles, and determines its area




```python
import math  # importing the math module

a = float(input("Enter first side length of triangle: "))
b = float(input("Enter second side length of triangle: "))
c = float(input("Enter third side length of triangle: "))

# Check whether the given sides can form a triangle
if (a + b > c) and (a + c > b) and (b + c > a):

    # Calculate the semi-perimeter
    s = (a + b + c) / 2

    # Calculate area using Heron's formula
    area = math.sqrt(s * (s - a) * (s - b) * (s - c))

    # Calculate angles using the Law of Cosines
    cos_A = (b**2 + c**2 - a**2) / (2 * b * c)
    cos_B = (a**2 + c**2 - b**2) / (2 * a * c)
    cos_C = (a**2 + b**2 - c**2) / (2 * a * b)

    # Convert angles from radians to degrees
    angle_A = math.degrees(math.acos(cos_A))
    angle_B = math.degrees(math.acos(cos_B))
    angle_C = math.degrees(math.acos(cos_C))

    # Display the results
    print(f"\nArea: {area:.2f} square units")
    print(f"Angle A (opposite to side a): {angle_A:.2f} degrees")
    print(f"Angle B (opposite to side b): {angle_B:.2f} degrees")
    print(f"Angle C (opposite to side c): {angle_C:.2f} degrees")

else:
    print("Invalid sides. These sides cannot form a triangle.")
```
