# if...else Statement

# if-else statements evaluate conditions to execute different blocks of code. They allow a program to make decisions based on whether a statement is True or False.

# Provides an alternative block of code if the initial condition fails.

```python
# SYNTAX

if condition:
    # Code to execute if condition is True
else:
    # Code to execute if condition is False
```

```python
# EXAMPLE - NUMBER IS EVEN OR ODD

number = int(input("enter a number: "))

if number % 2 == 0:
  print(number," is even number")
else:
  print(number, " is odd number")
```

# if...elif...else Statement

# you use it to check multiple conditions sequentially

# Python stops checking further conditions once it finds a match

```python
# SYNTAX

if condition1:
    # Code to execute if condition1 is True
elif condition2:
    # Code to execute if condition2 is True
elif condition3:
    # Code to execute if condition3 is True
else:
    # Code to execute if none of the above conditions are True
```

```python
# EXAMPLE - NUMBER IS ODD, EVEN OR ZERO

number = int(input("enter a number: "))

if number>0:
  print("number is positive")
elif number<0:
  print("number is negative")
else:
  print("number is zero")
```

```python
# grading in markesheet

marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 80:
    print("Grade B")
elif marks >= 70:
    print("Grade C")
else marks >= 40:
    print("Grade p")
else:
  print("fail")
```
