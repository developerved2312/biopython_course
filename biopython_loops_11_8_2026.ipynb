```python
# REVISION OF IF-ELSE CONDITIONAL STATEMENTS
# Python Program for Temperature Conversion

temperature = float(input("enter the value of temperature: "))
choice = input("enter choice from celsius or fahrenheit ").lower()

if choice == "celsius":
    # conversion of  Celsius to Fahrenheit
    fahrenheit = (temperature * 9 / 5) + 32
    print("Temperature in Fahrenheit: ", fahrenheit)

elif choice == "fahrenheit":
    # Conversion Fahrenheit to Celsius
    celsius = (temperature - 32) * 5 / 9
    print("Temperature in Celsius:", celsius)

else:
    # error handling - like wrong spelling or any other word entered
    print("Invalid choice. Please enter Celsius or Fahrenheit.")
```

```python
# Write a Python program that takes a string as input and uses a counter to count the number of characters.
text = input("Enter a string: ")
counter = 0
for letter in text:
    counter += 1
print("Number of characters:", counter)
```

# Write a Python program that takes an integer N as input from the user and calculates the factorial of the number

# n ! = n x (n-1) x (n-2) X ..............  x 2 x 1

# eg 4! = 4 x 3 x 2 x 1 = 24

```python
# factorial of a number

n = int(input("Enter a number: "))
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(factorial)
```

# While Loop in Python

<h2> A while loop is used when you want to repeatedly execute a block of code as long as a given condition is True.  </h2>



```python
# Check condition → Execute code → Update condition → Repeat
```

```python
#SYNTAX
while condition:
    # Code to execute
    statement
```

while is the keyword used to create the loop.

condition is a Boolean expression that evaluates to True or False.

The indented statements execute repeatedly while the condition remains True.

Python uses indentation to identify the body of the loop.

<b> <h1>main difference </h1> </b>

# A while loop is generally useful when the number of iterations is not known beforehand.

# A for loop is commonly used when iterating over a known sequence or range

```python
# Write a Python program to print numbers from 1 to 10 using a while loop.

# Initialize the counter
i = 1
# Repeat while i is less than or equal to 10
while i <= 10:
    print(i,end=",")
    # Increment the counter
    i = i + 1  # or you can write i += 1 it has a similar logic
```

```python
# Write a Python program to print numbers from 10 to 1 using a while loop.

# Initialize the counter
i = 10
# Repeat until i becomes less than 1
while i >= 1:
    print(i,end=", ")
    # Decrease the counter
    i = i - 1
```

```python
# Print even numbers from 2 to 10

# we will start from number 2
i = 2
# Continue until 10
while i <= 10:
    print(i,end=",")
    # Iincrease by 2 to obtain next even number like 2, 2+2=4, 4+2=6 and so on...
    i = i + 2
```

```python
# Multiplication table by taking input from user
n = int(input("Enter a number: "))
i = 1  # start from 1
# Print 10 multiples
while i <= 10:
    print(n * i)
    # Move to the next multiple
    i = i + 1
```
